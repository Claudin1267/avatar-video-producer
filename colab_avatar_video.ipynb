## ✅ CÉLULA 4 - Upload da Imagem
*(clique em "Choose Files" quando aparecer)*

```python
from google.colab import files
import cv2
import os

print("📸 Clique no botão abaixo")
print("Selecione sua imagem (PNG ou JPG)")
print("Dica: rosto bem claro e bem iluminado\n")

uploaded = files.upload()

if uploaded:
    filename = list(uploaded.keys())[0]
    
    # Salvar em local correto
    avatar_path = f'/content/avatar-producer/inputs/{filename}'
    
    # Mover arquivo
    !mv {filename} {avatar_path}
    
    # Verificar
    img = cv2.imread(avatar_path)
    if img is not None:
        h, w = img.shape[:2]
        print(f"\n✅ Imagem recebida: {filename}")
        print(f"   Dimensões: {w}x{h} pixels")
        print(f"   Caminho: {avatar_path}")
    else:
        print("❌ Erro ao ler imagem")
else:
    print("❌ Nenhuma imagem enviada")
```

## ✅ CÉLULA 6 - Escrever o Script
*(EDITE O TEXTO AQUI)*

```python
# ⚙️ EDITE ESTE TEXTO!
# Escreva o que você quer que o avatar fale

script = """
Olá! Bem-vindo ao Avatar Video Producer.
Meu nome é Avatar IA.
Posso falar em vídeos de até 10 segundos.
Tudo 100% grátis no Google Colab.
Vamos criar conteúdo incrível juntos!
""".strip()

# Mostrar o script
print("📝 Seu script:")
print("-" * 50)
print(script)
print("-" * 50)
print(f"\n📊 Total: {len(script.split())} palavras")
print(f"⏱️  Tempo estimado: {len(script.split()) / 60:.1f} segundos")
```

## ✅ CÉLULA 7 - GERAR VÍDEO 🎬
*(vai levar 1-3 minutos - AGUARDE)*

```python
import os

# Procurar imagem
inputs_dir = '/content/avatar-producer/inputs'
avatar_files = [f for f in os.listdir(inputs_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

if not avatar_files:
    print("❌ Nenhuma imagem encontrada em inputs/")
    print("   Volte à CÉLULA 4 e faça upload")
else:
    avatar_image_path = os.path.join(inputs_dir, avatar_files[0])
    output_path = '/content/avatar-producer/outputs/video_resultado.mp4'
    
    print("="*60)
    print("🎬 GERANDO VÍDEO")
    print("="*60)
    print(f"📸 Imagem: {avatar_files[0]}")
    print(f"📝 Script: {len(script.split())} palavras")
    print(f"\n⏳ Processando... (pode levar 1-3 minutos)")
    print(f"❌ NÃO FECHE ESTA ABA!\n")
    
    # Criar producer
    producer = AvatarVideoProducer(work_dir='/content/avatar-producer')
    
    # Gerar
    success = producer.produce(
        text=script,
        avatar_image_path=avatar_image_path,
        output_path=output_path,
        duration_max=10
    )
    
    if success:
        print("\n" + "="*60)
        print("✅ VÍDEO GERADO COM SUCESSO!")
        print("="*60)
    else:
        print("\n⚠️  Processamento concluído (veja logs acima)")
```

## ✅ CÉLULA 8 - Ver Vídeo
*(célula rápida)*

```python
from IPython.display import Video
import os

output_path = '/content/avatar-producer/outputs/video_resultado.mp4'

if os.path.exists(output_path):
    size_mb = os.path.getsize(output_path) / (1024*1024)
    print(f"✅ Vídeo pronto!")
    print(f"   Arquivo: video_resultado.mp4")
    print(f"   Tamanho: {size_mb:.1f}MB")
    print(f"\n🎬 Assistindo...\n")
    display(Video(output_path, width=400))
else:
    print("❌ Vídeo não encontrado")
    print("   Execute a CÉLULA 7 primeiro")
```
